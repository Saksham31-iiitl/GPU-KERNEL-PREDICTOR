```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

metrics_df = pd.read_csv(r'c:\Users\saksh\OneDrive\Desktop\AMD Projects\GPU_KERNEL_PREDICTOR\data\processed\model_evaluation_metrics.csv')

plt.figure(figsize=(10, 6))

plt.subplot(1, 3, 1)
sns.barplot(x=metrics_df.columns, y=metrics_df.loc[0])
plt.title('Evaluation Metrics')

plt.subplot(1, 3, 2)
sns.heatmap(metrics_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')

plt.subplot(1, 3, 3)
sns.pairplot(metrics_df)
plt.title('Pairplot of Metrics')

plt.tight_layout()
plt.show()
```
