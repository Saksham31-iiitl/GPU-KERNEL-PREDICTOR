```python
import os
import pandas as pd
from sklearn.preprocessing import StandardScaler
import joblib
```

```python
PROJECT_DIR = os.path.abspath("..")
PROC_DIR    = os.path.join(PROJECT_DIR, "data", "processed")
CLEAN_CSV   = os.path.join(PROJECT_DIR, "data", "processed", "kernels_clean.csv")

FEATURES_OUT = os.path.join(PROC_DIR, "X.csv")
TARGET_OUT   = os.path.join(PROC_DIR, "y.csv")
SCALER_OUT   = os.path.join(PROC_DIR, "scaler.pkl")

os.makedirs(PROC_DIR, exist_ok=True)
```

```python
if not os.path.exists(CLEAN_CSV):
    raise FileNotFoundError(
        f"Cannot find cleaned data at {CLEAN_CSV}.\n"
        "Please run Step 1 (Data Preprocessing) first to generate kernels_clean.csv."
    )
print("✅ Found cleaned data:", CLEAN_CSV)
```

```python
df = pd.read_csv(CLEAN_CSV)
print("Clean data shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
target_col = "ocl_kernel"               # change as needed
exclude    = ["device", "platform", "day", target_col]
feature_cols = [c for c in df.columns if c not in exclude]

X = df[feature_cols]
y = df[target_col]

print(f"Selected {len(feature_cols)} features, e.g.: {feature_cols[:5]}")
print(f"Predicting target: '{target_col}'")
```

```python
scaler   = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Feature means (first 5):", scaler.mean_[:5])
print("Feature scales (first 5):", scaler.scale_[:5])
```

```python
pd.DataFrame(X_scaled, columns=feature_cols).to_csv(FEATURES_OUT, index=False)
pd.DataFrame(y, columns=[target_col]).to_csv(TARGET_OUT,   index=False)
joblib.dump(scaler, SCALER_OUT)

print("Saved:")
print(" • Features →", FEATURES_OUT)
print(" • Target   →", TARGET_OUT)
print(" • Scaler   →", SCALER_OUT)
```
