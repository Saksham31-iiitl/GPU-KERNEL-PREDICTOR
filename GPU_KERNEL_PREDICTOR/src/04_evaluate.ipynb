```python
import os
import pandas as pd
import joblib
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

PROJECT_DIR = os.path.abspath(os.path.join(os.getcwd(), ".."))
SRC_DIR = os.path.join(PROJECT_DIR, "src")
PROC_DIR = os.path.join(PROJECT_DIR, "data", "processed")

X_CSV = os.path.join(SRC_DIR, "X.csv")
Y_CSV = os.path.join(SRC_DIR, "y.csv")
MODEL_OUT = os.path.join(PROC_DIR, "rf_model.pkl")

model = joblib.load(MODEL_OUT)
X_test = pd.read_csv(X_CSV)
y_test = pd.read_csv(Y_CSV).iloc[:, 0]

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("R²:", r2)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd

# Make predictions on the test data
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# Save evaluation metrics to a DataFrame
evaluation_metrics = {
    "MAE": [mae],
    "MSE": [mse],
    "R²": [r2]
}

metrics_df = pd.DataFrame(evaluation_metrics)

# Define the path to save the evaluation metrics
metrics_output = os.path.join(PROC_DIR, "model_evaluation_metrics.csv")

# Save the metrics to CSV
metrics_df.to_csv(metrics_output, index=False)

print(f"✅ Evaluation metrics saved to: {metrics_output}")
```
